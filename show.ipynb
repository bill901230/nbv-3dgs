```python
import open3d as o3d
import plotly.graph_objects as go
import numpy as np

# 讀取 PCD 檔案
pcd = o3d.io.read_point_cloud("data/house3k/1/0.pcd")

# 轉換為 NumPy 陣列
points = np.asarray(pcd.points)

# 創建 Plotly 3D 點雲圖
fig = go.Figure(data=[go.Scatter3d(
    x=points[:, 0], 
    y=points[:, 1], 
    z=points[:, 2], 
    mode='markers',
    marker=dict(
        size=2,  # 點大小
        color=points[:, 2],  # 以 Z 軸顏色為依據
        colorscale='Viridis',
        opacity=0.8
    )
)])

fig.update_layout(
    title="3D Point Cloud Visualization",
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z"
    )
)

fig.show()
```

```python
import numpy as np
import open3d as o3d

# 讀取 PCD
pcd = o3d.io.read_point_cloud("data/house3k/1/0.pcd")
points = np.asarray(pcd.points)

# 檢查 NaN / Inf
print("有 NaN?", np.isnan(points).any())
print("有 Inf?", np.isinf(points).any())

# 檢查數據範圍
print("最小值:", np.min(points, axis=0))
print("最大值:", np.max(points, axis=0))
print("平均值:", np.mean(points, axis=0))

print("X 範圍:", np.min(points[:,0]), " ~ ", np.max(points[:,0]))
print("Y 範圍:", np.min(points[:,1]), " ~ ", np.max(points[:,1]))
print("Z 範圍:", np.min(points[:,2]), " ~ ", np.max(points[:,2]))
```

```python
import numpy as np
import open3d as o3d

# 讀取 PCD
pcd = o3d.io.read_point_cloud("data/house3k/1/0.pcd")
points = np.asarray(pcd.points)

# 1. 檢查並過濾 NaN / Inf
points = points[~np.isnan(points).any(axis=1)]
points[np.isinf(points)] = 0

# 2. 檢查範圍，是否需要縮小
if np.max(points) > 100:
    points /= 1000  # 假設是毫米，轉換為公尺

# 3. 添加紅色
colors = np.ones((len(points), 3))
colors[:, 1:] = 0  # 變成紅色
pcd.colors = o3d.utility.Vector3dVector(colors)

# 4. 嘗試 ASCII 格式
o3d.io.write_point_cloud("data/house3k/1/0_ascii.pcd", pcd, write_ascii=True)

# 5. 繪製點雲
fig = o3d.visualization.draw_plotly([pcd])
fig.show()
```

```python
import open3d as o3d
import numpy as np

# 讀取 PCD
pcd = o3d.io.read_point_cloud("data/house3k/1/0.pcd")

# 降採樣（減少點雲數量）
pcd = pcd.voxel_down_sample(voxel_size=0.05)

# 添加顏色（紅色）
colors = np.ones((len(pcd.points), 3))  # 預設為白色
colors[:, 1:] = 0  # 設為紅色
pcd.colors = o3d.utility.Vector3dVector(colors)

# 直接顯示點雲（不要手動啟動 WebRTC）
o3d.visualization.draw([pcd])
```
